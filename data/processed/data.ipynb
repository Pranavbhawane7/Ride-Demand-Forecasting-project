```python
import pandas as pd
```

```python
# Read full dataset
data = pd.read_csv("yellow_tripdata_2016-03.csv")

# Take random sample
sample_data = data.sample(
    n=100000,
    random_state=42
)
```

```python
# Rename columns
sample_data.rename(
    columns={
        'VendorID': 'vendor_id',
        'tpep_pickup_datetime': 'pickup_datetime',
        'tpep_dropoff_datetime': 'dropoff_datetime',
        'RatecodeID': 'ratecode_id'
    },
    inplace=True
)

# Save sample CSV
sample_data.to_csv(
    "yellow_trip_data.csv",
    index=False
)
```
